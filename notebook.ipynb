```python
def sorting(num):
    n = len(num)
    for i in range(n):           
        for j in range(0, n-i-1):
            if num[j] > num[j+1]:
                num[j], num[j+1] = num[j+1], num[j]
    return num

num = [1,4,5,6,8,2,3,7]
print("before:",num)
print("after:",sorting(num))
#num.sort()
#print(num)
```
