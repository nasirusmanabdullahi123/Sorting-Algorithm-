```python
def sorting(num):
    for i in range(len(num)):           
        for j in range(0, len(num)-i-1):
            if num[j] > num[j+1]:
                num[j], num[j+1] = num[j+1], num[j]

num = [1,4,5,6,8,2,3,7]
#num.sort()
#print(num)
```

```python
sorting(num)
```
